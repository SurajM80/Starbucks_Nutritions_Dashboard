### Create the First Dimension Table

```python
dim_category = df[['Beverage_category']].drop_duplicates().reset_index(drop=True)

dim_category['Category_ID'] = range(1, len(dim_category)+1)

dim_category = dim_category[['Category_ID', 'Beverage_category']]

dim_category
```

```python
# df.head()
```

### Create the 2nd Dimension Table

```python
dim_beverage = df[['Beverage']].drop_duplicates().reset_index(drop=True)  #from that column, duplicates are removed and the index is reset.

dim_beverage['Beverage_ID'] = range(1, len(dim_beverage)+1)  #This creates range(1,34), because here we have 33 unique values, that why it is..!, Beverage_ID becomes surrogate keys.

dim_beverage = dim_beverage[['Beverage_ID', 'Beverage']]    #This just rearranges the columns

dim_beverage
```

### Create 3rd Dimention table 

```python
dim_prep = df[['Beverage_prep']].drop_duplicates().reset_index(drop=True)

dim_prep['Prep_ID'] = range(1, len(dim_prep)+1)

dim_prep = dim_prep[['Prep_ID', 'Beverage_prep']]

dim_prep
```

```python
df.head()
```

# Create a FACT Table 

```python
fact_nutrition = df.merge(
    dim_category,
    on='Beverage_category',
    how='left'
)

fact_nutrition = fact_nutrition.merge(
    dim_beverage,
    on='Beverage',
    how='left'
)

fact_nutrition = fact_nutrition.merge(
    dim_prep,
    on='Beverage_prep',
    how='left'
)

fact_nutrition.head()
```

```python
fact_nutrition = fact_nutrition.drop(
    columns=[
        'Beverage_category',
        'Beverage',
        'Beverage_prep'
    ]
)

fact_nutrition.head()
```
